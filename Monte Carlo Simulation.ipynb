# Strategy: Bet with a constant amount of money, even if you lose or win.

```python

def rollDice():
    roll = random.randint(1,100) # gives a value between [1,100]
    
    if (roll==100):
        return False
    elif(roll<100 and roll>50):
        return True
    else:
        return False
def simple_bettor(funds, intial_wager,wager_count):
    value = funds
    wager = intial_wager
    current_wager = 0
    while current_wager < wager_count:
        if rollDice():
            value = value + wager
            
        else:
            value = value - wager
        current_wager = current_wager + 1
        print ("Funds", value)
simple_bettor(10000,100,100)
```

Observation: Set number of games to be large and every better with the same strategy of playing large number of games goes broke eventually.

We now have to see the trajectory, to compare bettors and assessing the volatility of their strategy, since the volatility or the draw down in the strategy denotes some form of risk, or added risk the more volatile the strategy is. Here we have a set of bettors using the same strategy, but ending up at different points, we have to see how they got their, their trajectory.

```python
import matplotlib.pyplot as plt
import matplotlib
import random
def rollDice():
    roll = random.randint(1,100) # gives a value between [1,100]
    
    if (roll==100):
        return False
    elif(roll<100 and roll>50):
        return True
    else:
        return False
def simple_bettor(funds, intial_wager,wager_count):
    value = funds
    wager = intial_wager
    wX = []
    vY = []
    
    current_wager = 1
    while current_wager <= wager_count:
        if rollDice():
            value = value + wager
            wX.append(current_wager)
            vY.append(value)
        else:
            value = value - wager
            wX.append(current_wager)
            vY.append(value)
        
        current_wager = current_wager + 1
    if(value<0):
        print("broke")
    else:
        print("Funds=",value)
    plt.plot(wX,vY)
for i in range(100):
    simple_bettor(10000,100,100000)
plt.ylabel("Account Value")
plt.xlabel("Wager Count")
plt.show()
```

As you increase the games tending to a large number, everyone starts to go in loss.

Eventually everyone loses on a long term scale, no one makes profit.<br> It proves that even if you are taking a risk 1%, it will eventually catch up to you.

# Strategy: Bet by doubling the loss, once you win you revert back to initial wager

```python
import matplotlib.pyplot as plt
import matplotlib
import random
def rollDice():
    roll = random.randint(1,100) # gives a value between [1,100]
    
    if (roll==100):
        return False
    elif(roll<100 and roll>50):
        return True
    else:
        return False
def doubler_bettor(funds, initial_wager,wager_count):
    value = funds
    wager = initial_wager
    wX = []
    vY = []
    factor = 1
    current_wager = 1
    while current_wager <= wager_count:
        if rollDice():
            value = value + factor*wager
            factor = 1
            wX.append(current_wager)
            vY.append(value)
        else:
            
            if((value-factor*wager)<0):
                print("Broke")
                break
            else:
                value = value - factor*wager
                factor = factor*2
            wX.append(current_wager)
            vY.append(value)
        
        current_wager = current_wager + 1
    if(value>0):
        print("Funds after final wager_count =",value)
    plt.plot(wX,vY)
for i in range(100):
    doubler_bettor(10000,100,100000)
plt.ylabel("Account Value")
plt.xlabel("Wager Count")
plt.show()
```
